# 01 - Matplotlib Basics
**Goal:** make the core chart types with matplotlib: line, bar, scatter, histogram, pie, subplots, and customization.

Install once: `pip install matplotlib seaborn plotly`

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

np.random.seed(42)   # same random numbers every run
plt.rcParams['figure.figsize'] = (7, 4)
```

## 1. Line plot
Best for trends over time.

```python
x = np.linspace(0, 10, 100)
plt.plot(x, np.sin(x), label='sin(x)')
plt.plot(x, np.cos(x), label='cos(x)')
plt.title('Line Plot')
plt.xlabel('x')
plt.ylabel('y')
plt.legend()
plt.show()
```

## 2. Bar chart (vertical)
Best for comparing categories.

```python
products = ['Laptop', 'Phone', 'Tablet', 'Watch']
sales = [120, 200, 80, 60]

plt.bar(products, sales, color='steelblue')
plt.title('Sales by Product')
plt.ylabel('Units Sold')
plt.show()
```

## 3. Horizontal bar chart
Great when category names are long.

```python
plt.barh(products, sales, color='seagreen')
plt.title('Sales by Product (horizontal)')
plt.xlabel('Units Sold')
plt.show()
```

## 4. Grouped bar chart

```python
q1 = [100, 150, 60, 40]
q2 = [120, 200, 80, 60]
pos = np.arange(len(products))
w = 0.35

plt.bar(pos - w/2, q1, w, label='Q1')
plt.bar(pos + w/2, q2, w, label='Q2')
plt.xticks(pos, products)
plt.title('Q1 vs Q2 Sales')
plt.legend()
plt.show()
```

## 5. Stacked bar chart
Shows how parts add up to a total.

```python
plt.bar(products, q1, label='Q1')
plt.bar(products, q2, bottom=q1, label='Q2')
plt.title('Stacked Bar: Q1 + Q2')
plt.legend()
plt.show()
```

## 6. Scatter plot
Shows the relationship between two numbers.

```python
ad_spend = np.random.uniform(10, 100, 50)
revenue = ad_spend * 3 + np.random.normal(0, 25, 50)

plt.scatter(ad_spend, revenue, color='tomato')
plt.title('Ad Spend vs Revenue')
plt.xlabel('Ad Spend ($k)')
plt.ylabel('Revenue ($k)')
plt.show()
```

## 7. Bubble chart (scatter + size)

```python
size = np.random.uniform(20, 300, 50)
plt.scatter(ad_spend, revenue, s=size, alpha=0.5, c=revenue, cmap='viridis')
plt.colorbar(label='Revenue')
plt.title('Bubble Chart')
plt.show()
```

## 8. Histogram
Shows how values are distributed.

```python
ages = np.random.normal(35, 10, 500)
plt.hist(ages, bins=20, color='orchid', edgecolor='white')
plt.title('Customer Age Distribution')
plt.xlabel('Age')
plt.ylabel('Count')
plt.show()
```

## 9. Pie chart
Use only for a few categories that add to 100%.

```python
share = [40, 30, 20, 10]
plt.pie(share, labels=products, autopct='%1.0f%%', startangle=90)
plt.title('Revenue Share')
plt.show()
```

## 10. Donut chart (pie with a hole)

```python
plt.pie(share, labels=products, autopct='%1.0f%%', wedgeprops={'width': 0.4})
plt.title('Donut Chart')
plt.show()
```

## 11. Area chart

```python
months = np.arange(1, 13)
rev = np.cumsum(np.random.randint(5, 20, 12))
plt.fill_between(months, rev, alpha=0.4)
plt.plot(months, rev)
plt.title('Cumulative Revenue (Area)')
plt.xlabel('Month')
plt.show()
```

## 12. Box plot

```python
data = [np.random.normal(m, 10, 100) for m in (50, 60, 55)]
plt.boxplot(data, labels=['North', 'South', 'West'])
plt.title('Box Plot by Region')
plt.show()
```

## 13. Error bars

```python
means = [50, 60, 55]
errors = [5, 8, 4]
plt.errorbar(['North', 'South', 'West'], means, yerr=errors, fmt='o', capsize=5)
plt.title('Mean with Error Bars')
plt.show()
```

## 14. Subplots
Several charts in one figure.

```python
fig, axes = plt.subplots(2, 2, figsize=(10, 7))

axes[0, 0].plot(x, np.sin(x))
axes[0, 0].set_title('Line')

axes[0, 1].bar(products, sales)
axes[0, 1].set_title('Bar')

axes[1, 0].scatter(ad_spend, revenue)
axes[1, 0].set_title('Scatter')

axes[1, 1].hist(ages, bins=15)
axes[1, 1].set_title('Histogram')

plt.tight_layout()
plt.show()
```

## 15. Customization
Colors, line styles, markers, annotations, grid, and saving.

```python
plt.style.use('ggplot')   # try: 'seaborn-v0_8', 'dark_background'

plt.plot(months, rev, color='darkred', linestyle='--', marker='o', linewidth=2)
plt.annotate('Peak', xy=(12, rev[-1]), xytext=(9, rev[-1] - 10),
             arrowprops=dict(arrowstyle='->'))
plt.title('Customized Line Chart', fontsize=14)
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.grid(True)
plt.savefig('images/custom_line.png', dpi=150)   # save as image
plt.show()
plt.style.use('default')
```
