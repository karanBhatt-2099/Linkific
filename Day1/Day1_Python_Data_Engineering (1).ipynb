## 02 - Lists, tuples, sets, dictionaries
`scripts/02_collections.py`

```python
# Title: 02 - Lists, tuples, sets, dictionaries
# LIST - ordered, can be changed
fruits = ["apple", "banana", "mango"]
fruits.append("orange")
print("List:", fruits)
print("First item:", fruits[0])

# TUPLE - ordered, cannot be changed
point = (10, 20)
print("Tuple:", point)

# SET - no duplicates
numbers = {1, 2, 2, 3, 3, 3}
print("Set:", numbers)

# DICTIONARY - key : value
student = {"name": "Ravi", "age": 21}
student["city"] = "Mumbai"
print("Dictionary:", student)
print("Name:", student["name"])
```

## 03 - List and dict comprehensions
`scripts/03_comprehensions.py`

```python
# Title: 03 - List and dict comprehensions
numbers = [1, 2, 3, 4, 5]

# Normal way
squares = []
for n in numbers:
    squares.append(n * n)
print("Loop:", squares)

# List comprehension (short way)
squares = [n * n for n in numbers]
print("Comprehension:", squares)

# With a condition
evens = [n for n in numbers if n % 2 == 0]
print("Even numbers:", evens)

# Dict comprehension
square_dict = {n: n * n for n in numbers}
print("Dict:", square_dict)
```

## 04 - Functions and lambda functions
`scripts/04_functions_lambda.py`

```python
# Title: 04 - Functions and lambda functions
def add(a, b):
    return a + b

print(add(3, 4))

# Default value
def greet(name, greeting="Hello"):
    return greeting + ", " + name

print(greet("Asha"))
print(greet("Asha", "Namaste"))

# Lambda = one-line function
square = lambda x: x * x
print(square(5))

# Lambda used for sorting
names = ["banana", "kiwi", "apple"]
print(sorted(names, key=lambda x: len(x)))
```

## 05 - Exception handling (try / except / finally)
`scripts/05_exception_handling.py`

```python
# Title: 05 - Exception handling (try / except / finally)
try:
    result = 10 / 0
    print(result)
except ZeroDivisionError:
    print("Error: you cannot divide by zero")
finally:
    print("This line always runs")

print("---")

try:
    number = int("abc")
except ValueError:
    print("Error: 'abc' is not a number")

print("---")

try:
    print(5 + 5)
except:
    print("Something went wrong")
else:
    print("No error happened")
```

## 08 - CSV handling

```python
# Title: 08 - CSV handling
import os
import csv
import pandas as pd
os.makedirs("data", exist_ok=True)

# Write a CSV
with open("data/students.csv", "w", newline="") as f:
    writer = csv.writer(f)
    writer.writerow(["name", "marks"])
    writer.writerow(["Asha", 85])
    writer.writerow(["Ravi", 72])
    writer.writerow(["Meera", 91])

# Read with csv module
with open("data/students.csv", "r") as f:
    for row in csv.reader(f):
        print(row)

# Read with pandas (easier)
df = pd.read_csv("data/students.csv")
print(df)
print("Average marks:", df["marks"].mean())
```

## 13 - DateTime: parse and format dates
`scripts/13_datetime_parse_format.py`

```python
# Title: 13 - DateTime: parse and format dates
from datetime import datetime

# String -> date (parse)
text = "2024-03-15 14:30:00"
d = datetime.strptime(text, "%Y-%m-%d %H:%M:%S")
print("Parsed:", d)
print("Year:", d.year, "Month:", d.month, "Day:", d.day)

# Date -> string (format)
print(d.strftime("%d/%m/%Y"))
print(d.strftime("%B %d, %Y"))
print(d.strftime("%A"))

# Current date and time
print("Now:", datetime.now())
```

## 14 - DateTime: timezone handling
`scripts/14_timezone_handling.py`

```python
# Title: 14 - DateTime: timezone handling
from datetime import datetime
from zoneinfo import ZoneInfo

utc_time = datetime(2024, 3, 15, 9, 0, tzinfo=ZoneInfo("UTC"))
print("UTC     :", utc_time)

india = utc_time.astimezone(ZoneInfo("Asia/Kolkata"))
print("India   :", india)

newyork = utc_time.astimezone(ZoneInfo("America/New_York"))
print("New York:", newyork)
```

## 15 - DateTime: date calculations

```python
# Title: 15 - DateTime: date calculations
from datetime import datetime, timedelta

today = datetime(2024, 3, 15)

print("Tomorrow  :", today + timedelta(days=1))
print("Next week :", today + timedelta(weeks=1))
print("30 days ago:", today - timedelta(days=30))

# Difference between two dates
start = datetime(2024, 1, 1)
end = datetime(2024, 12, 25)
diff = end - start
print("Days between:", diff.days)
```

## 17 - Exercise: read multiple CSV files and combine
`scripts/17_ex_combine_csv_files.py`

```python
# Title: 17 - Exercise: read multiple CSV files and combine
import os
import glob
import pandas as pd
os.makedirs("data/sales", exist_ok=True)

# Create 3 sample files
pd.DataFrame({"product": ["A", "B"], "amount": [100, 200]}).to_csv("data/sales/jan.csv", index=False)
pd.DataFrame({"product": ["A", "C"], "amount": [150, 80]}).to_csv("data/sales/feb.csv", index=False)
pd.DataFrame({"product": ["B", "C"], "amount": [120, 90]}).to_csv("data/sales/mar.csv", index=False)

# Read all and combine
all_data = []
for file in glob.glob("data/sales/*.csv"):
    all_data.append(pd.read_csv(file))

combined = pd.concat(all_data, ignore_index=True)
print(combined)
print("Total by product:")
print(combined.groupby("product")["amount"].sum())
```

## 19 - Exercise: process a log file
`scripts/19_ex_process_log_files.py`

```python
# Title: 19 - Exercise: process a log file
import os
os.makedirs("data", exist_ok=True)

# Create a sample log file
with open("data/app.log", "w") as f:
    f.write("2024-03-15 09:00:01 INFO User logged in\n")
    f.write("2024-03-15 09:05:42 ERROR Database timeout\n")
    f.write("2024-03-15 09:10:13 INFO File uploaded\n")
    f.write("2024-03-15 09:15:00 ERROR Payment failed\n")
    f.write("2024-03-15 09:20:30 WARNING Disk almost full\n")

# Read and count log levels
info = error = warning = 0
with open("data/app.log", "r") as f:
    for line in f:
        parts = line.split()      # split into words
        level = parts[2]          # third word is the level
        if level == "INFO":
            info += 1
        elif level == "ERROR":
            error += 1
            print("ERROR LINE:", line.strip())
        elif level == "WARNING":
            warning += 1

print("INFO:", info, "| ERROR:", error, "| WARNING:", warning)
```

## 20 - Exercise: date manipulation
`scripts/20_ex_date_manipulation.py`

```python
# Title: 20 - Exercise: date manipulation
from datetime import datetime, timedelta

# 1. Calculate age
birth = datetime(1995, 8, 20)
today = datetime(2024, 3, 15)
age = (today - birth).days // 365
print("1) Age:", age)

# 2. What day of the week was it?
print("2) Day:", birth.strftime("%A"))

# 3. Date after 100 days
print("3) After 100 days:", (today + timedelta(days=100)).date())

# 4. Change format
text = "15-03-2024"
d = datetime.strptime(text, "%d-%m-%Y")
print("4) New format:", d.strftime("%Y/%m/%d"))

# 5. Is it a weekend?
if today.weekday() >= 5:
    print("5) Weekend")
else:
    print("5) Weekday")
```

## 21 - Exercise: error handling in file operations
`scripts/21_ex_file_error_handling.py`

```python
# Title: 21 - Exercise: error handling in file operations
import json

# 1. File does not exist
try:
    with open("data/missing_file.txt", "r") as f:
        print(f.read())
except FileNotFoundError:
    print("1) File not found!")

# 2. Bad JSON
try:
    json.loads("this is not json")
except json.JSONDecodeError:
    print("2) Invalid JSON!")

# 3. Bad number in data
values = ["10", "20", "abc", "40"]
total = 0
for v in values:
    try:
        total += int(v)
    except ValueError:
        print("3) Skipping bad value:", v)
print("Total:", total)
```
